```python
n = int(input('Введите количество строк: '))
triangle = []

for i in range(n):
    strochka = [] 
    for el in range(i + 1):
        if el == 0 or el == i:
            strochka.append(1)
        else:
            strochka.append(triangle[i-1][el-1] + triangle[i-1][el])     
    triangle.append(strochka)
    
last_strochka = ' '.join(map(str, triangle[-1]))
max_len = len(last_strochka)

for strochka in triangle:
    strochka_in_str = ' '.join(map(str, strochka))
    spaces = (max_len - len(strochka_in_str)) // 2
    print(' ' * spaces + strochka_in_str)
```
