```python
"""
3. Составить мультиагентную систему с тем же набором инструментов, но разделив модели на логически понятные роли. Можете даже Structured Output использовать
"""

import os
from dotenv import load_dotenv
from agno.agent import Agent
from agno.team import Team
from agno.tools.duckduckgo import DuckDuckGoTools
from agno.tools.reasoning import ReasoningTools
from agno.models.openai.like import OpenAILike

load_dotenv()

model = OpenAILike(
    id=os.getenv('MODEL_ID'),
    base_url=os.getenv('OPENROUTER_BASE_URL'),
    api_key=os.getenv('OPENROUTER_API_KEY')
)

researcher = Agent(
    name="Researcher",
    role="Search web for information",
    model=model,
    tools=[DuckDuckGoTools(enable_search=True, enable_news=True)],
    instructions="""
    You are a web researcher. Find relevant information using DuckDuckGo.
    - Search for multiple sources when needed
    - If web search fails, use your general knowledge
    - Be thorough but concise
    - Cite sources when available
    """,
    markdown=True,
)

analyst = Agent(
    name="Analyst",
    role="Analyze and structure information",
    model=model,
    tools=[ReasoningTools(add_instructions=True)],
    instructions="""
    You analyze information and plan comprehensive responses.
    - Use think() to identify ALL questions in the user's request
    - Make sure EVERY question gets answered
    - Structure data logically
    - Identify key points for each question
    - Connect related information
    """,
    markdown=True,
)

reviewer = Agent(
    name="Reviewer",
    role="Review and improve final response",
    model=model,
    tools=[ReasoningTools(add_instructions=True)],
    instructions="""
    You review and improve responses from other agents.
    - Use think() to verify ALL questions were answered
    - Check for completeness - did we miss any part of the request?
    - Ensure accuracy and quality
    - Structure the final answer clearly with sections
    - Add any missing information
    """,
    markdown=True,
)

search_team = Team(
    name="Multi-Agent Search Team",
    members=[researcher, analyst, reviewer],
    model=model,
    instructions="""
    You are a team of specialized agents working together.
    
    CRITICAL: Answer ALL questions in the user's request, even if there are multiple topics.
    
    Workflow:
    1. Analyst identifies ALL questions and plans comprehensive response
    2. Researcher finds information for EACH question (web search or knowledge)
    3. Reviewer verifies ALL questions are answered and improves the response
    
    Work efficiently and ensure nothing is missed.
    """,
    markdown=True,
)
```

```python
from IPython.display import display, Markdown

display(Markdown("Multi-Agent Search Team"))

query = "Привет! Я хочу приготовить пасту карбонара на ужин, дай мне простой рецепт. А кстати, я вообще очень люблю итальянскую кухню и часто готовлю пасту дома. Еще мне интересно узнать - какие есть полезные советы по здоровому питанию? И подскажешь, что лучше выбрать для изучения веб-разработки - Python или JavaScript?"

display(Markdown(f"Запрос:\n{query}"))

response = search_team.run(query, stream=False)

display(Markdown(f"Ответ команды:\n{response.content}"))
```
