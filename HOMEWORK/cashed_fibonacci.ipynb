```python
N = int(input("Enter a number to compute Fibonacci: "))
print('You entered:', N)

MY_CASH: dict[int, int] = {0: 0, 1: 1, 2: 1}
def fib_cached(N: int) -> int:
    if N in MY_CASH:
        return MY_CASH[N]
    MY_CASH[N] = fib_cached(N-1) + fib_cached(N-2)
    return MY_CASH[N]

print('Fibonacci (cached):', fib_cached(N))
```
