Запросить у пользователя - хочет ли он сняться в кино
Если да - предложить две разных роли - Халк или Локи
    Если халк - запросить сколько у него бицепс в объеме
        Если меньше 60 - отправить домой
        Еслиб больше или равно - сказать что он принят
    Если локи - спросить кого он больше любит - маму или папу
        Если папу - отправить домой
        Если маму - отправить к папе, чтобы спросил кого больше любит
Если нет - попрощаться
Во всех условиях обрабатывать белиберду и неправильные выборы ошибкой для пользователя

```python
from rich import print as rprint

rprint('[bold yellow]Welcome to the our casting show!')

print('Do u want to star in a movie? [YES/NO]: ')
decision: str = input()
rprint(decision)

if decision.lower() == 'yes':
    print('Do u want to play Hulk or Loki? write! : ')
    fav_role: str = input()
    print(fav_role)
    
    if fav_role.lower() == 'hulk':
        
        print("What is your biceps size? ")
        biceps_size: str = input()
        print(biceps_size)
        
        if biceps_size.isdigit():  
            biceps_size = int(biceps_size)  
            if biceps_size >= 60:
                rprint("[bold green]The role is YOURS![/]")
            elif biceps_size <= 0:
                rprint("[bold red]Crazy man..[/]")
        else:
            rprint("[bold blue]Go home, buddy[/]")
            
    elif fav_role.lower() == 'loki':
        
        print('who do u like more: mom or dad? answer! :')
        fav_parent: str = input()
        print(fav_parent)
        
        if fav_parent.lower() == 'mom':
            rprint('[bold blue]Go home and ask u ask dad who u like more')
        elif fav_parent.lower() == 'dad':
            rprint('[bold blue] Go home, buddy')
        else:
            rprint('[bold red]Crazy man..')
            
    else: 
        rprint('[bold italic]U\'re not serious enough, buddy![/bold italic] 🙃')
        
elif decision.lower() == 'no':
    rprint('[bold blue] Well, huh, goodbuy! 👋')
    
else: 
    rprint('[bold italic]U\'re not serious enough, buddy![/bold italic] 🙃')
```
