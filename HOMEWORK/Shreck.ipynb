 === Наследование базовых классов
    BaseCharacter --> Shrek
    BaseCharacter --> PussInBoots
    BaseCharacter --> Donkey
    BaseCharacter --> JackHorner

    BaseCharacter --> BaseInActionCharacter
    BaseInActionCharacter --> ShrekInAction
    BaseInActionCharacter --> PussInBootsInAction
    BaseInActionCharacter --> DonkeyInAction
    BaseInActionCharacter --> JackHornerInAction

    BaseCharacter --> BaseFunkoPop
    BaseFunkoPop --> ShrekFunkoPop
    BaseFunkoPop --> PussInBootsFunkoPop
    BaseFunkoPop --> DonkeyFunkoPop
    BaseFunkoPop --> JackHornerFunkoPop

    BaseCharacter --> BaseCosplayer
    BaseHuman --> BaseCosplayer
    BaseCosplayer --> ShrekCosplayer
    BaseCosplayer --> PussInBootsCosplayer
    BaseCosplayer --> DonkeyCosplayer
    BaseCosplayer --> JackHornerCosplayer

=== Применение миксинов (через реализацию) ===
    MixinSpeakable --> BaseCharacter
    MixinActionable --> BaseInActionCharacter
    MixinCollectible --> BaseFunkoPop
    MixinPoseable --> BaseCosplayer
    MixinCostumeWearable --> BaseCosplayer
    MixinAnimated --> BaseCharacter
    MixinFunny --> BaseCharacter

```python
class MixinSpeakable :
    def speak(self):
        return f"{self.name} говорит: '{self.catchphrase}'"

class MixinActionable :
    def perform_action(self):
        return f"{self.name} {self.signature_move}!"
    def stomp_foot(self):
        return f"{self.name} топает ногой - БУМ!"
    def roar(self):
        return f"{self.name} рычит: РРРАААА!"

class MixinCollectible :
    is_collectible: bool = True
    def collect(self):
        return f"Фигурка {self.name} добавлена в коллекцию"
    def display(self):
        return f"{self.name} стоит на полке"
    
class MixinPoseable :
    def pose(self):
        return f"{self.name} принимает героическую позу!"
    def strike_pose(self):
        return f"{self.name} делает свою фирменную позу"

class MixinCostumeWearable :
    def wear_costume(self):
        return f"*Надеваем костюм {self.name}*"
    def take_photo(self):
        return f"*Фото косплея {self.name} готово!*"
    def apply_makeup(self):
        return f"*Наносим грим для {self.name}*"

class MixinAnimated :
    is_animated: bool = True
    def animate(self):
        return f"{self.name} двигается в мультфильме"
    
class MixinFunny :
    def make_laugh(self):
        return f"{self.name}: {self.joke}"

class BaseHuman(MixinSpeakable):
    human_name : str
    is_animated : bool
    age : int
    hobby : str
    catchphrase : str

class BaseCharacter(MixinSpeakable, MixinAnimated, MixinFunny):
    name: str
    catchphrase: str
    joke: str

class BaseInActionCharacter(BaseCharacter, MixinActionable):
    action_mode = True
    signature_move = "атакует"
    quest : str

class BaseFunkoPop(BaseCharacter, MixinCollectible, MixinPoseable):
    is_vinyl_figure = True
    rarity : str
    pose_style : str
    
class BaseCosplayer(BaseCharacter, BaseHuman, MixinCostumeWearable, MixinPoseable):
    costume_quality = "профессиональный"
    props_count = 5
     
class Shrek(BaseCharacter):
    name = "Шрек"
    catchphrase = "Это МОЁ болото!"
    joke = "Огры как лук - у них много слоёв!"
    
class PussInBoots(BaseCharacter):
    name = "Кот в сапогах"
    catchphrase = "Посмотри в мои глаза!"
    joke = "Почему коты не играют в покер? Потому что боятся собак!"
    
class Donkey(BaseCharacter):
    name = "Осёл"
    catchphrase = "Мы едем в Венецию!"
    joke = "Почему ослы не летают? Потому что у них нет крыльев!"
     
class JackHorner(BaseCharacter):
    name = "Джек Хорнер"
    is_villain = True
    catchphrase = "Я заслуживаю волшебство!"
    joke = "Злодеи тоже умеют шутить!"
    
class ShrekInAction(Shrek, BaseInActionCharacter):
    quest = "спасти принцессу Фиону"
    signature_move = "отпугивает толпу рыком"
    
class PussInBootsInAction(PussInBoots, BaseInActionCharacter):
    quest = "помочь Шреку"
    signature_move = "делает милые глазки"
    
class DonkeyInAction(Donkey, BaseInActionCharacter):
    quest = "быть верным другом"
    signature_move = "отвлекает врагов разговорами"
    
class JackHornerInAction(JackHorner, BaseInActionCharacter):
    quest = "собрать все магические предметы"
    signature_move = "колдует с помощью магии" 

class ShrekFunkoPop(Shrek, BaseFunkoPop):
    rarity = "rare"
    pose_style = "стоя с поднятой рукой" 

class PussInBootsFunkoPop(PussInBoots, BaseFunkoPop):
    rarity = "common"
    pose_style = "стоя с поднятым мечом"
    
class DonkeyFunkoPop(Donkey, BaseFunkoPop):
    rarity = "uncommon"
    pose_style = "стоя с открытым ртом"
    
class JackHornerFunkoPop(JackHorner, BaseFunkoPop):
    rarity = "legendary"
    pose_style = "стоя с поднятой рукой"
    
class ShrekCosplayer(Shrek, BaseCosplayer):
    human_name = "Алексей"
    props_count = 3
    makeup_required = True
    green_bodysuit = True
    
class PussInBootsCosplayer(PussInBoots, BaseCosplayer):
    human_name = "Мария"
    props_count = 4
    requires_sword = True
    hat_with_feather = True
    
class DonkeyCosplayer(Donkey, BaseCosplayer):
    human_name = "Иван"
    props_count = 2
    donkey_ears = True
    tail_attached = True
    
class JackHornerCosplayer(JackHorner, BaseCosplayer):
    human_name = "Сергей"
    props_count = 5
    requires_magical_props = True
    bag_of_artifacts = True

if __name__ == "__main__":
    shrek = Shrek()
    print(shrek.speak())
    print(shrek.make_laugh())
    
    shrek_action = ShrekInAction()
    print(shrek_action.perform_action())
    print(shrek_action.stomp_foot())
    
    shrek_funko = ShrekFunkoPop()
    print(shrek_funko.collect())
    print(shrek_funko.pose())
    
    shrek_cosplayer = ShrekCosplayer()
    print(shrek_cosplayer.wear_costume())
    print(shrek_cosplayer.take_photo())
```
